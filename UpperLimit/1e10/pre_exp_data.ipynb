```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# 读取数据
file = "/data/home/zzh/data/Dataset_Filted/Experiment/1e10_V04/2024_all_merged_1e10_V04.npz"
data = np.load(file, allow_pickle=True)
print(data)
hitsE = data["hitsE"]; hitsM = data["hitsM"]
df = pd.DataFrame(data["params"], columns=data["param_names"])
indices=df[df['R_ue']<-2.5].index
df_filtered = df.loc[indices]
# 保存新的DataFrame到新的npz文件
hitsE_filtered = hitsE[indices]
hitsM_filtered = hitsM[indices]
np.savez("/data/home/zzh/data/Dataset_Filted/Experiment/1e10_V04/2024_all_merged_1e10_V04_filtered.npz",
         hitsE=hitsE_filtered, hitsM=hitsM_filtered, params=df_filtered.values, param_names=df_filtered.columns.values)
    
```
